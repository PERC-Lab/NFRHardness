```python
import json

path = "data_annotated.json"
with open(path, 'r', encoding='utf-8') as file:
    data = json.load(file)
```

```python
TYPES = ["Lexical", "Syntactic", "Semantic", "Vagueness", "Incompleteness", "Referential"]
count = {i:0 for i in TYPES}
for r in data:
    for t in TYPES:
        if r[t]:
            count[t]+=1
count
```

```python
import matplotlib.pyplot as plt

plt.bar(count.keys(), count.values(), color="steelblue")
plt.ylabel("Count")
plt.title("Ambiguity types")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
